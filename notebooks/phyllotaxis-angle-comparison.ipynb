# Phyllotaxis angle comparison

This notebook explains the phyllotaxis comparison card in this repo. The goal is simple: make the golden-angle choice legible instead of treating the sunflower picture as decorative magic.

The comparison uses the **Vogel model** for disk packing:

$$	heta_n = n\alpha, \qquad r_n = c\sqrt{n}$$

where $\alpha$ is the turn angle between successive seeds and $c$ is a scale factor.

The golden angle is

$$\alpha_g = 2\pi\left(1 - \frac{1}{\varphi}\right) = \pi(3-\sqrt{5}) \approx 137.507764^\circ.$$

The key idea is not mysticism. It is modular arithmetic: if $\alpha / 2\pi$ is a rational fraction $p/q$, then directions repeat after $q$ steps and visible spoke families form. If the turn is irrational, exact repeats never happen, and coverage of the disk is much more even.

```python
import math

golden_angle_deg = 180.0 * (3.0 - math.sqrt(5.0))
candidates = [
    ('3/8 turn', 135.0, 'repeat denominator 8'),
    ('golden angle', golden_angle_deg, 'irrational turn'),
    ('2/5 turn', 144.0, 'repeat denominator 5'),
]

for name, deg, note in candidates:
    print(f'{name:14s}  {deg:10.6f}°  {note}')
```

## Why the rational turns make spokes

For a rational turn $\alpha = 2\pi p/q$, the set of directions $n\alpha \pmod{2\pi}$ falls onto only $q$ residues. The radial factor $\sqrt{n}$ keeps pushing points outward, but the angular reuse makes the eye see spokes instead of a broadly reused disk.

That is why $3/8$ and $2/5$ still look organized even though they are close to the golden angle numerically. The closeness in degrees is not the main story. The rational-versus-irrational structure is.

```python
import statistics

SEEDS = 1400
SECTORS = 72

def sector_cv(angle_deg, seeds=SEEDS, sectors=SECTORS):
    counts = [0] * sectors
    angle = math.radians(angle_deg)
    for n in range(1, seeds + 1):
        theta = (n * angle) % (2.0 * math.pi)
        idx = min(sectors - 1, int(theta / (2.0 * math.pi) * sectors))
        counts[idx] += 1
    mean = sum(counts) / sectors
    return statistics.pstdev(counts) / mean, min(counts), max(counts)

for name, deg, _ in candidates:
    cv, lo, hi = sector_cv(deg)
    print(f'{name:14s}  CV={cv:0.3f}  min={lo:2d}  max={hi:3d}')
```

The sector coefficient of variation is only a rough proxy, but it is useful here. Lower values mean the angle reuses the disk more evenly across coarse angular bins. In this simple model, the golden angle wins decisively.

This is why the repo SVG now carries both a visual comparison and a small quantitative label.

## Caveats

- This is a **stylized packing model**, not a full theory of plant development.
- Even when the golden angle gives a cleaner disk-filling picture here, real botanical phyllotaxis also depends on growth dynamics and developmental constraints.
- The sector score is deliberately coarse. It is a supporting number, not a universal quality metric.
