# Rule 30 left-right asymmetry

This notebook is the slower companion to the new Rule 30 asymmetry card. The question is not whether Rule 30 looks chaotic in general. The question is whether the two flanks of the wedge behave the same once you measure matched diagonals the same depth in from each edge.


## The diagonal coordinates

If `s(t, x)` is the cell at step `t` and horizontal offset `x` from the seed, then the matched diagonals are

$$L_n(t) = s(t, -t + n), \qquad R_n(t) = s(t, t - n).$$

The depth `n` is the same on both sides. That keeps the left-versus-right comparison honest.


```python
from python.rule30_left_right_asymmetry import study_asymmetry

study = study_asymmetry(steps=320, max_depth=24, tail_window=192, max_period=64, block_size=5)
[(metric.side, metric.depth, metric.detected_period, round(metric.normalized_block_entropy, 4)) for metric in study.metrics[:8]]
```

## Main bounded result

- short tail periods appear at `24` of the first `24` left-edge depths but only `8` of the matched right-edge depths
- mean normalized `5`-bit entropy is `0.283` on the left versus `0.717` on the right
- mean transition rate is `0.479` on the left versus `0.736` on the right

That is a real directional split, not just a prettier rendering of the same wedge.

The top-right panel is the sharpest one. Small exact tail periods survive repeatedly on the left. The matched right diagonals mostly refuse them inside the same search window.


```python
selected = [metric for metric in study.metrics if metric.depth in (8, 16)]
[(metric.side, metric.depth, metric.detected_period, round(metric.transition_rate, 4), round(metric.normalized_block_entropy, 4)) for metric in selected]
```

## Two concrete depths

- at depth `8`, the left diagonal shows period `4` with entropy `0.400`, while the right diagonal shows period `32` with entropy `0.761`
- at depth `16`, the left diagonal shows period `4` with entropy `0.400`, while the right diagonal shows period `0` with entropy `0.859`

Those are not isolated cherry-picked oddities. They sit inside the broader pattern shown by the depth sweep.


## Caveats

1. This is a bounded finite-window audit, not a proof packet about infinite asymptotics.
2. The exact-period search only looks up to the configured limit, so `0` means "none found here," not a proof of aperiodicity.
3. Entropy is finite-block entropy on diagonal sequences, so it is a local complexity read rather than a universal chaos score.
